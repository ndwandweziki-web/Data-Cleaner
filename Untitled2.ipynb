```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# impoting ze data from ze csv
df = pd.read_csv('/Nat_Gas (1).csv')
df['Dates'] = pd.to_datetime(df['Dates'])
df = df.sort_values('Dates')

# doin some feature engineeing 4 ze modelz
df['Month'] = df['Dates'].dt.month
df['Year'] = df['Dates'].dt.year
df['Time'] = df['Year'] + (df['Month'] - 1) / 12
df['Sin_Month'] = np.sin(2 * np.pi * df['Month'] / 12)
df['Cos_Month'] = np.cos(2 * np.pi * df['Month'] / 12)

# trainin ze model 2 get the predictions
X = df[['Time', 'Sin_Month', 'Cos_Month']]
y = df['Prices']
model = LinearRegression().fit(X, y)

# function 2 get da price estimate 4 any date
def get_price(input_date_str):
    input_date = pd.to_datetime(input_date_str)
    month = input_date.month
    year = input_date.year

    time = year + (month - 1) / 12
    sin_month = np.sin(2 * np.pi * month / 12)
    cos_month = np.cos(2 * np.pi * month / 12)

    prediction = model.predict(np.array([[time, sin_month, cos_month]]))
    return prediction[0]

# extrapolatin 4 a year into the future
last_date = df['Dates'].max()
future_dates = pd.date_range(start=last_date + pd.DateOffset(months=1), periods=12, freq='M')
future_df = pd.DataFrame({'Dates': future_dates})
future_df['Month'] = future_df['Dates'].dt.month
future_df['Year'] = future_df['Dates'].dt.year
future_df['Time'] = future_df['Year'] + (future_df['Month'] - 1) / 12
future_df['Sin_Month'] = np.sin(2 * np.pi * future_df['Month'] / 12)
future_df['Cos_Month'] = np.cos(2 * np.pi * future_df['Month'] / 12)
future_df['Prices'] = model.predict(future_df[['Time', 'Sin_Month', 'Cos_Month']])

# plotzing
combined_df = pd.concat([df[['Dates', 'Prices']], future_df[['Dates', 'Prices']]])

plt.figure(figsize=(11, 5))
plt.plot(combined_df['Dates'], combined_df['Prices'], marker='o', linestyle='-', label='Price Series')
plt.axvline(x=pd.to_datetime('2024-09-30'), color='red', linestyle='--', label='Forecast Start')
plt.title('Natural Gas Price: Historical & 12-Month Forecast')
plt.xlabel('Date')
plt.ylabel('Price')
plt.grid(True)
plt.legend()
plt.show()
```

```python
def price_storage_contract(injection_dates, withdrawal_dates,
                           injection_volumes, withdrawal_volumes,
                           max_storage, injection_fee, withdrawal_fee,
                           transport_fee, rental_fee_monthly):
    """
    Calculates the value of the storage contract based on cash flows.
    """
    total_value = 0

    # Calculate revenue from sales and costs from purchases
    for i in range(len(injection_dates)):
        buy_price = get_price(injection_dates[i])
        total_value -= (injection_volumes[i] * buy_price)
        total_value -= (injection_volumes[i] * injection_fee)
        total_value -= (injection_volumes[i] * transport_fee)

    for i in range(len(withdrawal_dates)):
        sell_price = get_price(withdrawal_dates[i])
        total_value += (withdrawal_volumes[i] * sell_price)
        total_value -= (withdrawal_volumes[i] * withdrawal_fee)
        total_value -= (withdrawal_volumes[i] * transport_fee)

    # Calculating storage rental
    # Assuming storage duration is from first injection to last withdrawal
    start_date = pd.to_datetime(min(injection_dates))
    end_date = pd.to_datetime(max(withdrawal_dates))
    months_stored = (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month)
    total_value -= (months_stored * rental_fee_monthly)

    return total_value

# Example Test Case
inj_dates = ['2025-05-01', '2025-06-01']
wd_dates = ['2025-12-01', '2026-01-01']
volumes = [500000, 500000] # Total 1 million MMBtu

contract_value = price_storage_contract(
    inj_dates, wd_dates, volumes, volumes,
    max_storage=2000000, injection_fee=0.01, withdrawal_fee=0.01,
    transport_fee=0.05, rental_fee_monthly=50000
)

print(f"Contract Value: ${contract_value:,.2f}")
```
